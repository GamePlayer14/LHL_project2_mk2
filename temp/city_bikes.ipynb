Parse through the response to get the details you want for the bike stations in that city (latitude, longitude, number of bikes). 

```python
response.json()
```

```python
r = response.json()
q = r['network']

station_ids = [station["id"] for station in r["network"]["stations"]]
station_names = [station["name"] for station in r["network"]["stations"]]
latitudes = [station["latitude"] for station in r["network"]["stations"]]
longitudes = [station["longitude"] for station in r["network"]["stations"]]
free_bikes = [station["free_bikes"] for station in r["network"]["stations"]]
empty_slots = [station["empty_slots"] for station in r["network"]["stations"]]
```

Put your parsed results into a DataFrame.

```python
empty_slots
```

```python
av = []
print(type(empty_slots[0]))
for i1,i2 in zip(empty_slots,free_bikes):
    if type(i1) == class 'NoneType':
        sub = 0
        av.append((i2-sub)/i2)
    else:
        av.append((i2-i1)/i2)

```

```python
df = pd.DataFrame({
    "Station ID": station_ids,
    "Name": station_names,
    "Latitude": latitudes,
    "Longitude": longitudes,
    "Availability": (free_bikes-empty_slots)/free_bikes
})
df.to_csv("df_city_bikes.csv")
```
