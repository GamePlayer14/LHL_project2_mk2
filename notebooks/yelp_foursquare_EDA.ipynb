# Yelp

Send a request to Yelp with a small radius (1000m) for all the bike stations in your city of choice. 

```python
dfa2 = []
for _, row in df0.iloc[:150].iterrows():
    url2 = "https://api.yelp.com/v3/businesses/search?categories=restaurants,bars,landmarks&latitude={}&longitude={}&limit=50&radius=1000&sortBy=ratings".format(row['Latitude'], row['Longitude'])
    
    headers2 = {
        "Authorization": "Bearer ZwpyAgyaE0Ylhh1weYuF4gVGDbo2o83eUC-Y4VI1Z-XpxGMG2US-glIYZKNT2XWXfjPOhGo79pF_q48XzYHM8bo37J1E8G1y4U1wWNDBkhDm02XZgxhVDP1_uJ6aZ3Yx"
    }
    response = requests.get(url2, headers=headers2)

    dfa2.append(response.json())
```

Parse through the response to get the POI (such as restaurants, bars, etc) details you want (ratings, name, location, etc)

```python
dfa2[0]
```

```python
list = []
statNameList = df0['Name'][:150]
nameIndex = -1
for item in dfa2:
    nameIndex += 1
    for loca in item['businesses']:
        sl = []
        sl.append(loca['name'])
        sl.append(loca['distance'])
        sl.append(loca['rating'])
        sl.append(loca['review_count'])
        sl.append(loca['price'])
        sl.append(statNameList[nameIndex])
        list.append(sl)
```

Put your parsed results into a DataFrame

```python
df2 = pd.DataFrame(list, columns=['Name','Distance','Rating','Station Name'])
```
